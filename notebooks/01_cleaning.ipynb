```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../data/raw/hotel_bookings.csv')
print("Shape:", df.shape)
df.head()
```

```python
df.dtypes
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0]
print("Columns with missing values:\n")
print(missing)
```

```python
df['children'].fillna(0, inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['agent'].fillna(0, inplace=True)
df.drop(columns=['company'], inplace=True)

print("Missing values after cleaning:", df.isnull().sum().sum())
print("New shape:", df.shape)
```

```python
print("Duplicate rows:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Shape after removing duplicates:", df.shape)
```

```python
print("Cancellation distribution:\n")
print(df['is_canceled'].value_counts())
print("\nPercentage:\n")
print(df['is_canceled'].value_counts(normalize=True).mul(100).round(2))
```

```python
df.drop(columns=['reservation_status', 'reservation_status_date'], inplace=True)
print("Shape after dropping leaky columns:", df.shape)
```

```python
df.to_parquet('../data/processed/bookings_clean.parquet', index=False)
print("Saved successfully!")
```
