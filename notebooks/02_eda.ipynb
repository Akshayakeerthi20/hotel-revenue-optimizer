```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('../data/processed/bookings_clean.parquet')
print("Shape:", df.shape)
df.head()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='is_canceled', data=df, palette='Set2')
plt.title('Cancellation Distribution')
plt.xticks([0, 1], ['Not Cancelled', 'Cancelled'])
plt.xlabel('Booking Status')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('../outputs/figures/01_cancellation_distribution.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='hotel', hue='is_canceled', data=df, palette='Set2')
plt.title('Hotel Type vs Cancellation')
plt.xlabel('Hotel Type')
plt.ylabel('Count')
plt.legend(['Not Cancelled', 'Cancelled'])
plt.tight_layout()
plt.show()
```
